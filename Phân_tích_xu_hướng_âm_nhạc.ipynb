```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('spotify_alltime_top100_songs.csv')
```

```python
print(df.info())
```

```python
print(df.head())
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df
```

```python
top_artists = df.groupby('artist')['total_streams_billions'].sum().sort_values(ascending=False).head(10)
top_artists.plot(kind='bar', title='Top 10 nghệ sĩ có lượt xem nhiều nhất')

plt.show()
```

```python
sns.scatterplot(data=df, x='energy', y='total_streams_billions')
plt.title('Sự tương quan giữa Energy và Streams')
plt.show()
```

```python
# tinh trung binh danceability theo tung nam
trb=df.groupby('release_year')['danceability'].mean().reset_index()
sns.lineplot(data=trb, x='release_year', y='danceability')
plt.title("Trung bình danceability theo từng năm")
plt.show()
```
